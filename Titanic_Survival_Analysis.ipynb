# Titanic Survival Analysis
**Objective:** Analyze the Titanic dataset to understand survival patterns based on gender, passenger class, and age groups[cite: 4].


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset[cite: 4]
df = pd.read_csv("titanic.csv")

# Preview data[cite: 4]
print("Dataset Preview:")
print(df.head())
```

## 1. Data Cleaning
We need to clean the data by handling missing values, specifically in the 'Age' column[cite: 4].

```python
# Check missing values
print("Missing values before cleaning:\n", df.isnull().sum())

# Handle missing values in Age by filling with median
df['Age'] = df['Age'].fillna(df['Age'].median())

print("\nMissing values after cleaning:\n", df.isnull().sum())
```

## 2. Survival by Gender
Answering the question: Who survived more: males or females?[cite: 4]

```python
# Survival rate by gender[cite: 4]
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Survival Rate by Gender")
plt.show()

# Print actual numbers
print(df.groupby('Sex')['Survived'].mean() * 100)
```

## 3. Survival by Passenger Class
Answering the question: Did passenger class affect survival chances?[cite: 4]

```python
# Survival rate by passenger class[cite: 4]
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Survival Rate by Passenger Class")
plt.show()

# Print actual numbers
print(df.groupby('Pclass')['Survived'].mean() * 100)
```

## 4. Survival by Age Group
Answering the question: What was the survival rate by age group?[cite: 4] We will also look at the histogram of passenger ages[cite: 4].

```python
# Histogram of ages[cite: 4]
df["Age"].plot(kind="hist", bins=20, title="Passenger Ages", edgecolor='black')
plt.show()

# Create Age Groups and calculate survival
bins = [0, 12, 18, 35, 60, 100]
labels = ['Child', 'Teen', 'Young Adult', 'Adult', 'Senior']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

# Plot survival by Age Group
sns.barplot(x="AgeGroup", y="Survived", data=df)
plt.title("Survival Rate by Age Group")
plt.show()
```

## 5. Conclusion
* **Gender:** Females had a much higher survival rate compared to males.
* **Class:** 1st-class passengers had the highest survival chances, showing class significantly affected survival[cite: 4].
* **Age:** Children had higher survival rates compared to other age groups.
